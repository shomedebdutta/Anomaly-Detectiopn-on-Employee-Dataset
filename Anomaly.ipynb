```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('Employee_Attendance.csv')
```

```python
df.head(5)
```

```python
df['IsManual'] = df['IsManual'].astype('boolean')

mask_out = df['Present'] & df['OutTime'].isna()
df.loc[mask_out, 'OutTime'] = '20:00:00'
df.loc[mask_out, 'IsManual'] = True
df.loc[mask_out, 'OutDiff'] = 0
df.loc[mask_out, 'EarlyDeparture'] = False


mask_in = df['Present'] & df['InTime'].isna()
df.loc[mask_in, 'InTime'] = '10:30:00'
df.loc[mask_in, 'IsManual'] = True
df.loc[mask_in, 'InDiff'] = 0
df.loc[mask_in, 'IsLate'] = False
```

```python
import pandas as pd

expected_ip_map = {
    'M01': '192.168.1.10',
    'M03': '192.168.1.20',
    'M02': '192.168.1.15',
    'M04_HQ': '10.0.0.5'
}

df['ExpectedIP'] = df['MachineNo'].map(expected_ip_map)

invalid_rows = df[df['IPAddress'] != df['ExpectedIP']]
```

```python
df = df[df['Present'] == True]
```

```python
df
```

```python
df['InTime'] = pd.to_datetime(df['InTime'], format='%H:%M:%S', errors='coerce')
df['OutTime'] = pd.to_datetime(df['OutTime'], format='%H:%M:%S', errors='coerce')

target_in = pd.to_datetime('10:30:00', format='%H:%M:%S')
target_out = pd.to_datetime('20:00:00', format='%H:%M:%S')

filte_rows = df[

    (df['IsManual'] == True) &
    (df['InTime'] == target_in) &
    (df['OutTime'] == target_out)
]
```

```python
filte_rows
```

```python
df['row_id'] = df.index
filte_rows = df[
    (df['IsManual'] == True) &
    (df['InTime'] == target_in) &
    (df['OutTime'] == target_out)
].copy()

df['IsAnomaly'] = 0

df.loc[df['row_id'].isin(filte_rows['row_id']), 'IsAnomaly'] = 1

df.drop(columns='row_id', inplace=True)
```

```python
df['IsAnomaly'].value_counts()
```

```python
df['InTime'] = pd.to_datetime(df['InTime'])
df['OutTime'] = pd.to_datetime(df['OutTime'])
df['InSeconds'] = df['InTime'].dt.hour * 3600 + df['InTime'].dt.minute * 60 + df['InTime'].dt.second
df['OutSeconds'] = df['OutTime'].dt.hour * 3600 + df['OutTime'].dt.minute * 60 + df['OutTime'].dt.second
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['EmpId'] = le.fit_transform(df['EmpId'])
df['Present'] = le.fit_transform(df['Present'])
df['IsManual'] = le.fit_transform(df['IsManual'])
df['MachineNo'] = le.fit_transform(df['MachineNo'])
df['IsLate'] = le.fit_transform(df['IsLate'])
df['EarlyDeparture'] = le.fit_transform(df['EarlyDeparture'])
```

```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
```

```python
df.sample(2)
```
